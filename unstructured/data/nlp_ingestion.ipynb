```python
for case_id, text in zip(df["case_id"], df["text"]):

    for person in canonical_people:

        patterns = [
            person + " uses phone ",
            person + " used phone ",
            person + " using phone "
        ]

        for pattern in patterns:
            if pattern in text:
                start = text.index(pattern) + len(pattern)
                phone = re.match(r"\d{10}", text[start:])

                if phone:
                    relationships.append({
                        "case_id": case_id,
                        "source": person,
                        "relation": "USES",
                        "target": phone.group()
                    })

relationships_df = pd.DataFrame(relationships)

print("Total relationships:", len(relationships_df))
print("\nRelationship types:")
print(relationships_df["relation"].value_counts())
```

```python
relationships = []

for case_id, text in zip(df["case_id"], df["text"]):

    # CONTACTED / CALLED / MET
    for person1 in canonical_people:
        for person2 in canonical_people:
            if person1 != person2:

                for verb, relation in [
                    ("contacted", "CONTACTED"),
                    ("called", "CALLED"),
                    ("met", "MET")
                ]:
                    pattern = person1 + " " + verb + " " + person2

                    if pattern in text:
                        relationships.append({
                            "case_id": case_id,
                            "source": person1,
                            "relation": relation,
                            "target": person2
                        })

    # PERSON -> USES -> PHONE
    sentences = re.split(r"[.!?]", text)

    for sentence in sentences:
        phones = re.findall(r"\b\d{10}\b", sentence)

        if not phones:
            continue

        # Direct "uses/used phone"
        for person in canonical_people:
            if re.search(
                rf"\b{re.escape(person)}\s+(uses|used)\s+phone\s+\d{{10}}\b",
                sentence
            ):
                phone = re.search(r"\b\d{10}\b", sentence).group()

                relationships.append({
                    "case_id": case_id,
                    "source": person,
                    "relation": "USES",
                    "target": phone
                })

        # "Person contacted/called/met ... using phone"
        for person in canonical_people:
            if re.search(
                rf"^{re.escape(person)}\s+(contacted|called|met)\b.*\busing phone\s+\d{{10}}\b",
                sentence.strip()
            ):
                phone = re.search(r"\b\d{10}\b", sentence).group()

                relationships.append({
                    "case_id": case_id,
                    "source": person,
                    "relation": "USES",
                    "target": phone
                })

relationships_df = pd.DataFrame(relationships)

print("Total relationships:", len(relationships_df))
print("\nRelationship types:")
print(relationships_df["relation"].value_counts())
```

```python
# Convert relationships into the team's common schema

relationship_type_map = {
    "USES": "USED",
    "OWNS": "OWNS",
    "CALLED": "CALLED",
    "CONTACTED": "ASSOCIATED_WITH",
    "MET": "ASSOCIATED_WITH"
}

final_relationships_df = relationships_df.copy()

final_relationships_df["relationship_type"] = (
    final_relationships_df["relation"]
    .map(relationship_type_map)
)

# Add relationship IDs
final_relationships_df["relationship_id"] = [
    f"R{i:03d}" for i in range(1, len(final_relationships_df) + 1)
]

# Rename columns to match team schema
final_relationships_df = final_relationships_df.rename(columns={
    "source": "source_name",
    "target": "target_name"
})

# Source record = case/report where relationship was found
final_relationships_df["source_record"] = final_relationships_df["case_id"]

# Prototype confidence
final_relationships_df["confidence"] = 0.95

# Fields not yet available from the reports
final_relationships_df["date"] = ""
final_relationships_df["properties"] = "{}"

# Keep only the final schema columns for now
final_relationships_df = final_relationships_df[
    [
        "relationship_id",
        "source_name",
        "relationship_type",
        "target_name",
        "date",
        "source_record",
        "confidence",
        "properties"
    ]
]

print("Final relationships:", len(final_relationships_df))
final_relationships_df.head(10)
```

```python
# Create final entity IDs

entity_prefix = {
    "PERSON": "P",
    "LOCATION": "L",
    "PHONE": "PH",
    "VEHICLE": "V",
    "BANK_ACCOUNT": "BA",
    "ORGANIZATION": "ORG"
}

# Get unique entities and their source records
final_entities_df = (
    entities_df
    .groupby(["entity_type", "entity_value"])["case_id"]
    .apply(lambda x: ";".join(sorted(set(x))))
    .reset_index(name="source_record")
)

# Sort entities by type and name
final_entities_df = final_entities_df.sort_values(
    ["entity_type", "entity_value"]
).reset_index(drop=True)

# Generate IDs within each entity type
final_entities_df["number"] = (
    final_entities_df.groupby("entity_type").cumcount() + 1
)

final_entities_df["entity_id"] = final_entities_df.apply(
    lambda row: f"{entity_prefix[row['entity_type']]}{row['number']:03d}",
    axis=1
)

# Final schema fields
final_entities_df["name"] = final_entities_df["entity_value"]

final_entities_df["normalized_name"] = (
    final_entities_df["name"]
    .str.lower()
    .str.strip()
)

# First-name alias for people
final_entities_df["aliases"] = final_entities_df.apply(
    lambda row: row["name"].split()[0]
    if row["entity_type"] == "PERSON"
    else "",
    axis=1
)

final_entities_df["confidence"] = 0.95
final_entities_df["properties"] = "{}"

# Arrange columns according to team schema
final_entities_df = final_entities_df[
    [
        "entity_id",
        "entity_type",
        "name",
        "normalized_name",
        "aliases",
        "source_record",
        "confidence",
        "properties"
    ]
]

print("Total unique entities:", len(final_entities_df))
print("\nEntity counts:")
print(final_entities_df["entity_type"].value_counts())

final_entities_df.head(15)
```

```python
# Create a lookup from entity name → entity ID

entity_lookup = dict(
    zip(final_entities_df["name"], final_entities_df["entity_id"])
)

# Copy relationship table
final_relationships_with_ids = final_relationships_df.copy()

# Convert source and target names into entity IDs
final_relationships_with_ids["source_id"] = (
    final_relationships_with_ids["source_name"]
    .map(entity_lookup)
)

final_relationships_with_ids["target_id"] = (
    final_relationships_with_ids["target_name"]
    .map(entity_lookup)
)

# Arrange according to the team's final relationship schema
final_relationships_with_ids = final_relationships_with_ids[
    [
        "relationship_id",
        "source_id",
        "relationship_type",
        "target_id",
        "date",
        "source_record",
        "confidence",
        "properties"
    ]
]

print("Total relationships:", len(final_relationships_with_ids))
print("\nMissing source IDs:",
      final_relationships_with_ids["source_id"].isna().sum())

print("Missing target IDs:",
      final_relationships_with_ids["target_id"].isna().sum())

final_relationships_with_ids.head(10)
```

```python
# Validate relationship types against the team's common schema

allowed_relationships = {
    "LIVES_AT",
    "ASSOCIATED_WITH",
    "INVOLVED_IN",
    "CALLED",
    "OWNS",
    "USED",
    "LOCATED_AT",
    "REGISTERED_TO",
    "VISITED",
    "MEMBER_OF"
}

# Check for unsupported relationship types
invalid_relationships = set(
    final_relationships_with_ids["relationship_type"]
) - allowed_relationships

print("Invalid relationship types:", invalid_relationships)

# Check for missing IDs
print(
    "Missing source IDs:",
    final_relationships_with_ids["source_id"].isna().sum()
)

print(
    "Missing target IDs:",
    final_relationships_with_ids["target_id"].isna().sum()
)

# Final relationship counts
print("\nRelationship counts:")
print(
    final_relationships_with_ids["relationship_type"]
    .value_counts()
)

print("\nTotal relationships:",
      len(final_relationships_with_ids))
```
